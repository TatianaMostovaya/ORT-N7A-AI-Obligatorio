```python
import gymnasium as gym
from random import randint
```

#### Cargo el entorno

Sugerimos mantener `max_episode_steps = 700` para tener episodios más largos para entrenar.

```python
env = gym.make('Acrobot-v1', render_mode='human', max_episode_steps=700)
#env = gym.make('Acrobot-v1', render_mode='rgb_array', max_episode_steps=700) 
```

#### Simulación

```python
obs,_ = env.reset()
done = False
while not done:
    action = randint(0,2)
    obs, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated
    print(obs)
    env.render()
env.close()
```
